```python
import sys
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

BASE_DIR = Path.cwd().parent

if str(BASE_DIR) not in sys.path:
    sys.path.insert(0, str(BASE_DIR))

DATA_PATH = BASE_DIR / "data" / "document_risk_dataset.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Columns:")

for column in df.columns:
    print(column)
```

```python
print(df.info())
```

```python
print("Missing values:")

display(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())

print(
    "Duplicate documents:",
    df["text"].duplicated().sum()
)
```

```python
print("Risk categories:")

display(
    df["risk_category"].value_counts()
)
```

```python
print("Policy areas:")

display(
    df["policy_area"].value_counts()
)
```

```python
category_counts = (
    df["risk_category"]
    .value_counts()
)

plt.figure(figsize=(8, 5))

plt.bar(
    category_counts.index,
    category_counts.values
)

plt.xlabel("Risk Category")
plt.ylabel("Number of Documents")
plt.title("Risk Category Distribution")

plt.tight_layout()
plt.show()
```

```python
policy_counts = (
    df["policy_area"]
    .value_counts()
)

plt.figure(figsize=(10, 5))

plt.bar(
    policy_counts.index,
    policy_counts.values
)

plt.xlabel("Policy Area")
plt.ylabel("Number of Documents")
plt.title("Policy Area Distribution")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
df["text_length"] = (
    df["text"]
    .astype(str)
    .str.len()
)

display(
    df["text_length"].describe()
)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["text_length"],
    bins=20
)

plt.xlabel("Document Text Length")
plt.ylabel("Number of Documents")
plt.title("Document Length Distribution")

plt.tight_layout()
plt.show()
```

```python
print("Dataset summary")

print(
    "Total documents:",
    len(df)
)

print(
    "Risk categories:",
    df["risk_category"].nunique()
)

print(
    "Policy areas:",
    df["policy_area"].nunique()
)

print(
    "Average text length:",
    round(df["text_length"].mean(), 2)
)
```
